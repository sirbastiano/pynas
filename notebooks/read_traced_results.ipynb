```python
from pathlib import Path 

gen =2
txt_files = Path(f"/Data_large/marine/PythonProjects/OtherProjects/lpl-PyNas/models_traced/generation_{gen}")
txt_files = [x for x in txt_files.iterdir() if (x.is_file() and x.name.endswith('txt'))]

    # read the txt file
def read_test_iou(file_path: Path) -> float:
    """
    Reads the test_iou value from a txt file.

    Args:
        file_path (Path): Path to the txt file.

    Returns:
        float: The test_iou value.
    """
    with file_path.open('r') as file:
        for line in file:
            if line.startswith('test_iou:'):
                value = float(line.split(':')[1].strip())
                return value
    raise ValueError(f'No test_iou found in {file_path}')

# Collect all test_iou values
test_ious = [(f, read_test_iou(f)) for f in txt_files]
assert test_ious, 'No test_iou values found.'

# Find the file with the highest test_iou
top_file, top_iou = max(test_ious, key=lambda x: x[1])
print(f'Top test_iou: {top_iou} in file: {top_file}')

```

```python
import shutil 
# delete folder 

def delete_folder(folder_path: Path) -> None:
    """
    Deletes a folder and all its contents.
    
    Args:
        folder_path (Path): Path to the folder to delete.
    """
    if folder_path.exists() and folder_path.is_dir():
        shutil.rmtree(folder_path)
        print(f'Deleted folder: {folder_path}')
    else:
        print(f'Folder does not exist: {folder_path}')
        
        
delete_folder(Path('/Data_large/marine/PythonProjects/OtherProjects/lpl-PyNas/Results/backups/gen_0'))
```
