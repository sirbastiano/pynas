```python
from phidown import CopernicusDataSearcher
from phidown.ais import download_ais_data
```

```python
# ----------------------------------------------------------------------
# QUERY COPERNICUS FOR SENTINEL-1 PRODUCTS
# ----------------------------------------------------------------------

# Define area of interest (Antwerp region, Belgium) in WKT format
aoi_wkt = None

# Configure the search time window (full day by default)
target_date = "2025-08-22"  # YYYY-MM-DD format
start_time = "00:00:00"
start = f"{target_date}T{start_time}"
end_time = "23:00:00"
end_date = "2025-10-28"
end = f"{end_date}T{end_time}"

# Query Copernicus for Sentinel-1 Level-1 products over the AOI and date range
searcher = CopernicusDataSearcher()
searcher.query_by_filter(
    collection_name='SENTINEL-1',
    product_type=None,
    orbit_direction=None,
    cloud_cover_threshold=None,
    aoi_wkt=aoi_wkt,
    start_date=start,
    end_date=end,
    top=1000,
    attributes={'operationalMode': 'SM','processingLevel': 'LEVEL0'},
)
df = searcher.execute_query()
df
```

```python
# ----------------------------------------------------------------------
# SENSING WINDOW AND FOOTPRINT TO DOWNLOAD AIS MESSAGES
# ----------------------------------------------------------------------
# Extract timing metadata and footprint WKT from the product
sel_prod = df.iloc[-30]  # Select the last product from the search results

content_date = sel_prod['ContentDate']
target_date = content_date['Start'].split('T')[0]
start_time = content_date['Start'].split('T')[1]
end_time = content_date['End'].split('T')[1]
aoi_wkt = sel_prod['Footprint'].split(";")[1].replace("'", "")

# Download AIS data that overlaps the Sentinel-1 sensing window and footprint
try:
    # Request AIS tracks filtered by both time window and area of interest
    ais_data = download_ais_data(
        start_date=target_date,
        start_time=start_time,  # Handles Z suffix and microseconds correctly
        end_time=end_time,      # Matches Sentinel-1 sensing period precisely
        aoi_wkt=aoi_wkt
    )

    if not ais_data.empty:
        print(f"Actual data timespan: {ais_data['timestamp'].iloc[0]} to {ais_data['timestamp'].iloc[-1]}")
    else:
        print("No AIS detections matched the selected filters.")

except Exception as e:
    print(f"AIS download failed: {e}")
    print("Make sure you have installed: pip install phidown[ais]")

# Inspect the filtered AIS dataset
ais_data.head()
```

```python
from datetime import datetime, timedelta

def retrive_ais_data(sel_prod, buffer=30):
    """Retrieve AIS data for a given Sentinel-1 product with time buffer.
    
    Args:
        sel_prod (pandas.Series): Sentinel-1 product metadata containing ContentDate and Footprint.
        buffer (int): Time buffer in seconds to extend the search window before and after the sensing period.
    
    Returns:
        pandas.DataFrame or None: AIS data overlapping with the product's sensing window and footprint,
                                 or None if no data found or error occurred.
    """
    
    # Extract timing metadata and footprint WKT from the product
    content_date = sel_prod['ContentDate']
    target_date = content_date['Start'].split('T')[0]
    
    # Parse start and end times and apply buffer
    start_datetime = datetime.fromisoformat(content_date['Start'].replace('Z', '+00:00'))
    end_datetime = datetime.fromisoformat(content_date['End'].replace('Z', '+00:00'))
    
    # Apply buffer
    buffered_start = start_datetime - timedelta(seconds=buffer)
    buffered_end = end_datetime + timedelta(seconds=buffer)
    
    start_time = buffered_start.strftime('%H:%M:%S.%fZ')
    end_time = buffered_end.strftime('%H:%M:%S.%fZ')
    aoi_wkt = sel_prod['Footprint'].split(";")[1].replace("'", "")

    # Download AIS data that overlaps the Sentinel-1 sensing window and footprint
    try:
        # Request AIS tracks filtered by both time window and area of interest
        ais_data = download_ais_data(
            start_date=target_date,
            start_time=start_time,
            end_time=end_time,
            aoi_wkt=aoi_wkt
        )

        if not ais_data.empty:
            print(f'Actual data timespan: {ais_data["timestamp"].iloc[0]} to {ais_data["timestamp"].iloc[-1]}')
            return ais_data
        else:
            print('No AIS detections matched the selected filters.')
            return None

    except Exception as e:
        print(f'AIS download failed: {e}')
        print('Make sure you have installed: pip install phidown[ais]')
        return None
```
