```python
import pandas as pd 
from pathlib import Path 
import torch 
from pytorch_lightning import Trainer, LightningModule, LightningDataModule
from pynas.core.population import GenericLightningSegmentationNetwork, GenericLightningNetwork
import pytorch_lightning as pl
from IPython.display import clear_output

task = 'classification'  # or 'segmentation'

import warnings 
warnings.filterwarnings("ignore", category=UserWarning)  # To ignore any user warnings from libraries
from datasets.RawClassifier.loader import ClassifierDataModule
root_dir = '/Data_large/marine/PythonProjects/OtherProjects/lpl-PyNas/data/end2end'
dm = ClassifierDataModule(root_dir, batch_size=8, num_workers=0, transform=None, 
                    mode='semisplit' )

# define trainer
trainer = Trainer(max_epochs=1, enable_checkpointing=False, enable_progress_bar=False)


# =========================================
for gen_n in range(10):
    df = pd.read_pickle(f'/Data_large/marine/PythonProjects/OtherProjects/lpl-PyNas/Results_store/E2E/run2/Results/src/df_population_{gen_n}.pkl')
    for idx in range(15):
        model_path = f'/Data_large/marine/PythonProjects/OtherProjects/lpl-PyNas/Results_store/E2E/run2/models_traced/generation_{gen_n}/model_and_architecture_{idx}.pt'
        model = torch.jit.load(model_path, map_location=torch.device('cuda' if torch.cuda.is_available() else 'cpu'))
        model.eval()

        if task == 'segmentation':
            LM = GenericLightningSegmentationNetwork(
                                model=model,
                                learning_rate=0.01)
        else:
            LM = GenericLightningNetwork(
                                model=model,
                                learning_rate=0.01,
                                num_classes=2)

        # =========== Trainer Setup ===========
        trainer = pl.Trainer(
            accelerator='gpu',
            devices=1,
            max_epochs=1,
            callbacks=None,
            logger=False,  # Disable logging for cleaner output
            enable_checkpointing=False,  # Disable checkpointing for speed
        )

        results = trainer.test(LM, dm)
        test_mcc = results[0]['test_mcc']
        print(f'Test MCC: {test_mcc}')
        df.at[idx, 'test_mcc'] = test_mcc
    df.to_pickle(f'/Data_large/marine/PythonProjects/OtherProjects/lpl-PyNas/Results_store/E2E/run2/Results/src/df_population_{gen_n}.pkl')
    clear_output(wait=True)
```
